## Phase 5

### Dimensional Modeling

```python
df.head()
```

```python
# CREATE DIM_PRODUCT
dim_product = df[['ProductID']].drop_duplicates().sort_values('ProductID').reset_index(drop=True)
```

```python
dim_product
```

```python
# CREATE DIM_DATE 
dim_date = df[['DateKey']].drop_duplicates().reset_index(drop=True)
```

```python
dim_date
```

```python
# Force the column to be a string first, then convert using the exact YYYYMMDD structure
dim_date['Date'] = pd.to_datetime(dim_date['DateKey'].astype(str), format='%Y%m%d', errors='coerce')

dim_date['DayName'] = dim_date['Date'].dt.day_name()
dim_date['MonthName'] = dim_date['Date'].dt.month_name()
dim_date['Quarter'] = 'Q' + dim_date['Date'].dt.quarter.astype(str)
dim_date['Year'] = dim_date['Date'].dt.year
```

```python
dim_date
```

```python
# CREATE FACT_SALES
fact_sales = df[['SalesID', 'DateKey', 'PharmacyID', 'ProductID', 'UnitsSold', 'RevenueEUR', 'CostEUR', 'MarginEUR', 'PromoFlag']]
```

```python
fact_sales
```

```python
#  Export files as CSV file
dim_product.to_csv('DimProduct.csv', index=False)
dim_date.to_csv('DimDate.csv', index=False)
fact_sales.to_csv('FactSales.csv', index=False)
```
